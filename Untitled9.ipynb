```python
3a)arr = [21, 10, 45, 32, 67, 5, 18]
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

print(linear_search(arr, 32))
```

```python
b)def binary_search(arr, target):
    arr.sort()
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1

print(binary_search(arr, 45))
```

```python
c)def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr

print(bubble_sort(arr))
```

4a) class Node:
    def _init_(self, data):

b)class LinkedList:
    def _init_(self):
        self.head = None

    # Insert at beginning
    def insert_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # Insert at end
    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = new_node

    # Display list
    def display(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")

5a) queue
Justification: A queue follows the FIFO (First In, First Out) principle, meaning the first student to join the line is the first one to be served.

```python
5b) from collections import deque

queue = deque()

# Add student
def add_student(name):
    queue.append(name)

# Serve student
def serve_student():
    if queue:
        print("Served:", queue.popleft())
    else:
        print("Queue is empty")

# Display queue
def display_queue():
    print("Current Queue:", list(queue))

# Example
add_student("John")
add_student("Mary")
add_student("Peter")

display_queue()

serve_student()

display_queue()
```
